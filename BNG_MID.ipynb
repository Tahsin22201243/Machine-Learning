```python
#Practice: 01
sentence = "বাংলা ভাষা আমাদের পরিচয়ের অংশ।"

# 1. Print the sentence
print(sentence)

# 2. Print its Python data type
print(type(sentence))

# 3. Print its length
print(len(sentence))

# 4. Check whether "বাংলা" occurs in it
print("বাংলা" in sentence)

# 5. Replace "পরিচয়ের" with "সংস্কৃতির"
new_sentence = sentence.replace("পরিচয়ের", "সংস্কৃতির")
print(new_sentence)

# 6. Split the sentence using whitespace
words = sentence.split()
print(words)
```

```python
#Practice: 02

text = "রহিম বাংলা পড়ে। করিম ইংরেজি পড়ে। রহিম কবিতা পড়ে।"

# 1. Remove the punctuation mark "।"
cleaned_text = text.replace("।", "")
print(cleaned_text)

# 2. Split the cleaned text
words = cleaned_text.split()
print(words)

# 3. Count the total number of words
total_words = len(words)
print("Total words:", total_words)

# 4. Count the number of unique words
unique_words = set(words)
print("Unique words:", len(unique_words))

# 5. Create a frequency dictionary
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

print("Frequency:", frequency)
```

### 5.1 Script and Unicode

Bangla uses:

- vowel signs or কার,
- hasanta or হসন্ত,
- conjunct consonants or যুক্তাক্ষর,
- multiple Unicode code points that may appear as one visual unit.

Example:

- `ক + ি = কি`
- conjuncts: `ক্ষ`, `জ্ঞ`, `ত্র`

```python
examples = ["কি", "ক্ষ", "জ্ঞ", "ত্র"]

for word in examples:
    print(word, "-> Python length:", len(word))
```

### 5.2 Morphological richness

One root can produce many forms.

Verb root:

- কর
- করি
- করে
- করেছে
- করছিলাম

Noun forms:

- বই
- বইটি
- বইগুলো
- বইগুলোর


```python
verb_forms = ["কর", "করি", "করে", "করেছে", "করছিলাম"]
noun_forms = ["বই", "বইটি", "বইগুলো", "বইগুলোর"]

print("Verb forms:", verb_forms)
print("Noun forms:", noun_forms)
```

### Common Unicode categories

Some useful general categories are:

- `Lo` — Letter, Other
- `Mn` — Mark, Nonspacing
- `Mc` — Mark, Spacing Combining
- `Nd` — Decimal Digit Number
- `Po` — Punctuation, Other
- `Cf` — Format character

Do not memorize every category. The purpose is to see that a computer stores linguistic symbols with technical properties.


## 9. Unicode and UTF-8 are not the same thing

Unicode assigns an identity to a character.

UTF-8 determines how that identity is stored as bytes.

The same Unicode text may be encoded using different encodings, but UTF-8 is the dominant encoding for modern web and software systems.
